```python
"""
AI4I 2020 Predictive Maintenance Dataset
ベイズロジスティック回帰（線形）+ 層化K分割CV + MCMC診断・可視化
NumPyroバックエンド + tune=2000, draws=2000, chains=4
HDFに中心化済み交互作用項 (TempDiff × rpm) を追加
（Google Colab + Google Drive 版）
"""

import os
import numpy as np
import pandas as pd
import pymc as pm
import arviz as az
import matplotlib.pyplot as plt
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    roc_auc_score, average_precision_score, brier_score_loss,
    roc_curve, precision_recall_curve, confusion_matrix
)
from sklearn.calibration import calibration_curve
import warnings
warnings.filterwarnings("ignore")


# =============================================================
# 0. 出力ディレクトリ + サンプラー設定
# =============================================================
OUT_DIR = "/content/drive/MyDrive/Predictive Maintenance/outputs"
os.makedirs(OUT_DIR, exist_ok=True)
print(f"Output directory (persistent): {OUT_DIR}")

SAMPLER_KWARGS = dict(
    draws=2000,
    tune=2000,
    chains=4,
    target_accept=0.95,
    random_seed=42,
    progressbar=False,
    idata_kwargs={"log_likelihood": False},
    nuts_sampler="numpyro",
)
print(f"Sampler: {SAMPLER_KWARGS['nuts_sampler']}, "
      f"draws={SAMPLER_KWARGS['draws']}, tune={SAMPLER_KWARGS['tune']}, "
      f"chains={SAMPLER_KWARGS['chains']}")


# =============================================================
# 1. データ読み込みと特徴量エンジニアリング
# =============================================================
def load_and_prepare(path):
    df = pd.read_csv(path)
    df["TempDiff"]   = df["Process temperature [K]"] - df["Air temperature [K]"]
    df["Power"]      = df["Torque [Nm]"] * df["Rotational speed [rpm]"] * 2 * np.pi / 60
    df["PowerDev"]   = np.abs(df["Power"] - 6250)
    df["WearTorque"] = df["Tool wear [min]"] * df["Torque [Nm]"]
    df["TypeL"] = (df["Type"] == "L").astype(int)
    df["TypeM"] = (df["Type"] == "M").astype(int)
    df["TypeH"] = (df["Type"] == "H").astype(int)
    df["WearTorque_L"] = df["WearTorque"] * df["TypeL"]
    df["WearTorque_M"] = df["WearTorque"] * df["TypeM"]
    df["WearTorque_H"] = df["WearTorque"] * df["TypeH"]

    # ★ HDFの交互作用項: 中心化してから掛け算（共線性回避）
    td_centered  = df["TempDiff"] - df["TempDiff"].mean()
    rpm_centered = df["Rotational speed [rpm]"] - df["Rotational speed [rpm]"].mean()
    df["TempDiff_x_rpm"] = td_centered * rpm_centered

    return df


# =============================================================
# 2. ベイズロジスティック回帰
# =============================================================
def fit_bayes_logreg(X_train, y_train, prior_means=None, prior_sds=None,
                     feat_names=None):
    X_train = np.asarray(X_train, dtype=float)
    mu_X = X_train.mean(0)
    sd_X = X_train.std(0) + 1e-8
    Xs = (X_train - mu_X) / sd_X
    n, p = Xs.shape
    if prior_means is None: prior_means = np.zeros(p)
    if prior_sds   is None: prior_sds   = np.ones(p) * 2.0
    if feat_names  is None: feat_names  = [f"x{i}" for i in range(p)]

    coords = {"feature": feat_names}
    with pm.Model(coords=coords) as model:
        intercept = pm.Normal("intercept", mu=-5.0, sigma=2.0)
        beta = pm.Normal("beta", mu=prior_means, sigma=prior_sds, dims="feature")
        logit_p = intercept + pm.math.dot(Xs, beta)
        pm.Bernoulli("y_obs", logit_p=logit_p, observed=y_train)
        idata = pm.sample(**SAMPLER_KWARGS)
    return idata, (mu_X, sd_X)


def predict_proba(idata, X_new, scaler):
    mu_X, sd_X = scaler
    Xs = (np.asarray(X_new, dtype=float) - mu_X) / sd_X
    post = idata.posterior
    intercept = post["intercept"].values.reshape(-1)
    beta = post["beta"].values.reshape(-1, Xs.shape[1])
    logits = intercept[:, None] + beta @ Xs.T
    probs = 1.0 / (1.0 + np.exp(-logits))
    return {
        "mean":    probs.mean(0),
        "lower":   np.quantile(probs, 0.025, axis=0),
        "upper":   np.quantile(probs, 0.975, axis=0),
        "samples": probs,
    }


def fit_rnf_beta(y_train):
    """RNF用: 切片のみのベータ-ベルヌーイモデル"""
    with pm.Model():
        p_rnf = pm.Beta("p_rnf", alpha=1, beta=1)
        pm.Bernoulli("y", p=p_rnf, observed=y_train)
        idata = pm.sample(
            draws=2000, tune=1000, chains=4, random_seed=42,
            progressbar=False, idata_kwargs={"log_likelihood": False},
            nuts_sampler="numpyro",
        )
    return idata


# =============================================================
# 3. 故障モードごとの設定
# =============================================================
FEAT_MAP = {
    "TWF": {"feats": ["Tool wear [min]"],
            "prior_means": [3.0], "prior_sds": [2.0]},
    "HDF": {"feats": ["TempDiff", "Rotational speed [rpm]", "TempDiff_x_rpm"],
            "prior_means": [-3.0, -3.0, 2.0],   # 交互作用は正 (両方低いと高リスク)
            "prior_sds":   [2.0, 2.0, 2.0]},
    "PWF": {"feats": ["PowerDev"],
            "prior_means": [3.0], "prior_sds": [2.0]},
    "OSF": {"feats": ["WearTorque_L", "WearTorque_M", "WearTorque_H"],
            "prior_means": [4.0, 3.0, 2.0], "prior_sds": [2.0, 2.0, 2.0]},
    "RNF": {"feats": None, "prior_means": None, "prior_sds": None},
}


# =============================================================
# 4. 層化K分割CV
# =============================================================
def cross_validate_target(df, target, config, n_splits=5, random_state=42):
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    fold_metrics = []
    oof_pred  = np.zeros(len(df))
    oof_lower = np.zeros(len(df))
    oof_upper = np.zeros(len(df))
    y_all = df[target].values

    for fold, (tr_idx, te_idx) in enumerate(skf.split(df, y_all), start=1):
        tr = df.iloc[tr_idx]; te = df.iloc[te_idx]
        y_tr = tr[target].values; y_te = te[target].values

        if target == "RNF":
            idata = fit_rnf_beta(y_tr)
            p_post = idata.posterior["p_rnf"].values.reshape(-1)
            p_mean = np.full(len(te), p_post.mean())
            p_low  = np.full(len(te), np.quantile(p_post, 0.025))
            p_up   = np.full(len(te), np.quantile(p_post, 0.975))
        else:
            feats = config["feats"]
            idata, scaler = fit_bayes_logreg(
                tr[feats].values, y_tr,
                prior_means=config["prior_means"],
                prior_sds=config["prior_sds"],
                feat_names=feats,
            )
            pred = predict_proba(idata, te[feats].values, scaler)
            p_mean = pred["mean"]; p_low = pred["lower"]; p_up = pred["upper"]

        oof_pred[te_idx]  = p_mean
        oof_lower[te_idx] = p_low
        oof_upper[te_idx] = p_up

        n_pos = int(y_te.sum())
        if 0 < n_pos < len(y_te):
            auc = roc_auc_score(y_te, p_mean)
            ap  = average_precision_score(y_te, p_mean)
        else:
            auc = ap = np.nan
        brier = brier_score_loss(y_te, p_mean)

        fold_metrics.append({
            "target": target, "fold": fold, "n_test": len(y_te),
            "n_pos": n_pos, "AUC": auc, "AP": ap, "Brier": brier,
        })
        print(f"  [{target}] fold {fold}/{n_splits}: "
              f"n_pos={n_pos:3d}, AUC={auc:.3f}, AP={ap:.3f}, Brier={brier:.5f}")

    return pd.DataFrame(fold_metrics), oof_pred, oof_lower, oof_upper


# =============================================================
# 5. 最適しきい値
# =============================================================
def find_best_threshold(y_true, y_prob, criterion="f1"):
    if criterion == "f1":
        prec, rec, thr = precision_recall_curve(y_true, y_prob)
        prec, rec = prec[:-1], rec[:-1]
        f1 = 2 * prec * rec / (prec + rec + 1e-12)
        best_idx = int(np.argmax(f1))
        return thr[best_idx], {"F1": f1[best_idx],
                               "Precision": prec[best_idx],
                               "Recall": rec[best_idx]}
    elif criterion == "youden":
        fpr, tpr, thr = roc_curve(y_true, y_prob)
        j = tpr - fpr
        best_idx = int(np.argmax(j))
        return thr[best_idx], {"TPR": tpr[best_idx],
                               "FPR": fpr[best_idx], "J": j[best_idx]}


# =============================================================
# 6. MCMC診断・可視化
# =============================================================
def diagnose_and_visualize(idata, target, var_names, out_dir):
    summary = az.summary(idata, var_names=var_names, hdi_prob=0.95)
    print(f"\n[{target}] Posterior summary")
    print(summary)
    summary.to_csv(f"{out_dir}/posterior_summary_{target}.csv")

    rhat_max = summary["r_hat"].max()
    ess_min  = summary["ess_bulk"].min()
    print(f"  → Max R-hat = {rhat_max:.3f}  (要 <1.01)")
    print(f"  → Min ESS   = {ess_min:.0f}   (要 >400)")
    if rhat_max >= 1.01:
        print("  ⚠️  R-hat が 1.01 以上：チェーンが収束していない可能性あり")
    if ess_min < 400:
        print("  ⚠️  ESS が 400 未満：実効サンプル数が不足")

    az.plot_trace(idata, var_names=var_names, compact=True)
    plt.suptitle(f"Trace plot: {target}", y=1.01)
    plt.tight_layout()
    plt.savefig(f"{out_dir}/trace_{target}.png", dpi=120, bbox_inches="tight")
    plt.show()

    az.plot_posterior(idata, var_names=var_names, hdi_prob=0.95)
    plt.suptitle(f"Posterior distributions: {target}", y=1.01)
    plt.tight_layout()
    plt.savefig(f"{out_dir}/posterior_{target}.png", dpi=120, bbox_inches="tight")
    plt.show()

    if "beta" in var_names:
        az.plot_forest(idata, var_names=["beta"], hdi_prob=0.95, combined=True)
        plt.title(f"Coefficients (95% HDI): {target}")
        plt.axvline(0, color="red", linestyle="--", alpha=0.5)
        plt.tight_layout()
        plt.savefig(f"{out_dir}/forest_{target}.png", dpi=120, bbox_inches="tight")
        plt.show()

    az.plot_autocorr(idata, var_names=var_names, max_lag=50)
    plt.suptitle(f"Autocorrelation: {target}", y=1.01)
    plt.tight_layout()
    plt.savefig(f"{out_dir}/autocorr_{target}.png", dpi=120, bbox_inches="tight")
    plt.show()

    if "beta" in var_names:
        post = idata.posterior
        n_beta = post["beta"].shape[-1]
        if n_beta <= 4:
            az.plot_pair(idata, var_names=["intercept", "beta"],
                         kind="kde", marginals=True)
            plt.suptitle(f"Pair plot: {target}", y=1.01)
            plt.tight_layout()
            plt.savefig(f"{out_dir}/pair_{target}.png", dpi=120, bbox_inches="tight")
            plt.show()

    return summary


def fit_full_and_diagnose(df, target, config, out_dir):
    print(f"\n{'='*60}\n[{target}] Full-data fit & MCMC diagnostics\n{'='*60}")
    if target == "RNF":
        idata = fit_rnf_beta(df[target].values)
        var_names = ["p_rnf"]
        scaler = None
    else:
        feats = config["feats"]
        idata, scaler = fit_bayes_logreg(
            df[feats].values, df[target].values,
            prior_means=config["prior_means"],
            prior_sds=config["prior_sds"],
            feat_names=feats,
        )
        var_names = ["intercept", "beta"]
    summary = diagnose_and_visualize(idata, target, var_names, out_dir)
    return idata, scaler, summary


# =============================================================
# 7. メイン処理
# =============================================================
def main(csv_path, out_dir=OUT_DIR):
    os.makedirs(out_dir, exist_ok=True)

    print("=" * 70); print("Loading data...")
    df = load_and_prepare(csv_path)
    print(f"  Total samples: {len(df)}")
    for t in ["TWF","HDF","PWF","OSF","RNF","Machine failure"]:
        print(f"    {t:18s}: {df[t].mean()*100:.2f}%  (n_pos={int(df[t].sum())})")

    print("\n" + "=" * 70)
    print("Stratified K-Fold Cross-Validation per failure mode")
    print("=" * 70)
    all_fm = []; oof_predictions = {}
    for target, config in FEAT_MAP.items():
        print(f"\n--- {target} ---")
        fm, oof_m, oof_lo, oof_up = cross_validate_target(df, target, config)
        all_fm.append(fm)
        oof_predictions[target] = {"mean": oof_m, "lower": oof_lo, "upper": oof_up}
    fold_df = pd.concat(all_fm, ignore_index=True)

    # ---- Machine failure を fold ごとに集計 ----
    p_none = np.ones(len(df))
    for t in ["TWF","HDF","PWF","OSF","RNF"]:
        p_none *= (1.0 - oof_predictions[t]["mean"])
    p_machine = 1.0 - p_none
    y_m = df["Machine failure"].values

    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    machine_fold_rows = []
    for fold, (tr_idx, te_idx) in enumerate(skf.split(df, y_m), start=1):
        y_te = y_m[te_idx]
        p_te = p_machine[te_idx]
        n_pos = int(y_te.sum())
        auc = roc_auc_score(y_te, p_te) if 0 < n_pos < len(y_te) else np.nan
        ap  = average_precision_score(y_te, p_te) if n_pos > 0 else np.nan
        brier = brier_score_loss(y_te, p_te)
        machine_fold_rows.append({
            "target": "Machine failure", "fold": fold, "n_test": len(y_te),
            "n_pos": n_pos, "AUC": auc, "AP": ap, "Brier": brier,
        })
    machine_fold_df = pd.DataFrame(machine_fold_rows)
    fold_df = pd.concat([fold_df, machine_fold_df], ignore_index=True)

    print("\n" + "=" * 70); print("CV Summary"); print("=" * 70)
    summary = fold_df.groupby("target").agg(
        AUC_mean=("AUC","mean"),    AUC_std=("AUC","std"),
        AP_mean=("AP","mean"),      AP_std=("AP","std"),
        Brier_mean=("Brier","mean"),Brier_std=("Brier","std"),
        n_pos_total=("n_pos","sum"),
    ).round(4)
    print(summary)

    print("\n" + "=" * 70)
    print("Metrics per fold (including Machine failure)")
    print("=" * 70)
    print("\n=== AUC per fold ===")
    print(fold_df.pivot(index="target", columns="fold", values="AUC").round(4))
    print("\n=== AP per fold ===")
    print(fold_df.pivot(index="target", columns="fold", values="AP").round(4))
    print("\n=== Brier per fold ===")
    print(fold_df.pivot(index="target", columns="fold", values="Brier").round(5))

    print("\n" + "=" * 70); print("Machine failure (OR-combined, full OOF)"); print("=" * 70)
    print(f"  AUC   = {roc_auc_score(y_m, p_machine):.4f}")
    print(f"  AP    = {average_precision_score(y_m, p_machine):.4f}")
    print(f"  Brier = {brier_score_loss(y_m, p_machine):.5f}")

    print("\n" + "=" * 70); print("Optimal F1 thresholds (OOF)"); print("=" * 70)
    thr_rows = []
    for t in ["TWF","HDF","PWF","OSF","RNF"]:
        y, prob = df[t].values, oof_predictions[t]["mean"]
        if y.sum() == 0: continue
        thr, m = find_best_threshold(y, prob, criterion="f1")
        y_pred = (prob >= thr).astype(int)
        tn, fp, fn, tp = confusion_matrix(y, y_pred).ravel()
        thr_rows.append({"target": t, "threshold": float(thr),
                         "TP": tp, "FP": fp, "FN": fn, "TN": tn,
                         "Precision": m["Precision"], "Recall": m["Recall"],
                         "F1": m["F1"]})
    thr_df = pd.DataFrame(thr_rows); print(thr_df.to_string(index=False))

    thr_m, m_m = find_best_threshold(y_m, p_machine, criterion="f1")
    y_pred_m = (p_machine >= thr_m).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_m, y_pred_m).ravel()
    print(f"\nMachine failure: threshold={thr_m:.4f}, "
          f"TP={tp}, FP={fp}, FN={fn}, TN={tn}, "
          f"P={m_m['Precision']:.3f}, R={m_m['Recall']:.3f}, F1={m_m['F1']:.3f}")

    fold_df.to_csv(f"{out_dir}/cv_fold_metrics.csv", index=False)
    summary.to_csv(f"{out_dir}/cv_summary.csv")
    thr_df.to_csv(f"{out_dir}/optimal_thresholds.csv", index=False)
    pred_out = pd.DataFrame({"Machine failure": y_m, "p_machine_oof": p_machine})
    for t in ["TWF","HDF","PWF","OSF","RNF"]:
        pred_out[f"p_{t}_mean"]  = oof_predictions[t]["mean"]
        pred_out[f"p_{t}_lower"] = oof_predictions[t]["lower"]
        pred_out[f"p_{t}_upper"] = oof_predictions[t]["upper"]
    pred_out.to_csv(f"{out_dir}/oof_predictions.csv", index=False)
    print(f"\nSaved CSVs to: {out_dir}")

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    for t in ["TWF","HDF","PWF","OSF","RNF"]:
        y = df[t].values; p = oof_predictions[t]["mean"]
        if y.sum() == 0: continue
        fpr, tpr, _ = roc_curve(y, p)
        axes[0].plot(fpr, tpr, label=f"{t} (AUC={roc_auc_score(y,p):.3f})")
        prec, rec, _ = precision_recall_curve(y, p)
        axes[1].plot(rec, prec, label=f"{t} (AP={average_precision_score(y,p):.3f})")
    axes[0].plot([0,1],[0,1],"k--",alpha=0.3)
    axes[0].set_xlabel("FPR"); axes[0].set_ylabel("TPR"); axes[0].set_title("ROC (OOF)"); axes[0].legend()
    axes[1].set_xlabel("Recall"); axes[1].set_ylabel("Precision"); axes[1].set_title("PR (OOF)"); axes[1].legend()
    plt.tight_layout(); plt.savefig(f"{out_dir}/cv_curves.png", dpi=120); plt.show()

    fig, ax = plt.subplots(figsize=(6, 6))
    frac, mp = calibration_curve(y_m, p_machine, n_bins=10, strategy="quantile")
    ax.plot(mp, frac, "o-", label="Machine failure (OOF)")
    ax.plot([0,1],[0,1],"k--",alpha=0.4)
    ax.set_xlabel("Mean predicted prob"); ax.set_ylabel("Fraction of positives")
    ax.set_title("Calibration"); ax.legend()
    plt.tight_layout(); plt.savefig(f"{out_dir}/calibration.png", dpi=120); plt.show()

    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    for ax, metric in zip(axes, ["AUC", "AP", "Brier"]):
        fold_df.boxplot(column=metric, by="target", ax=ax)
        ax.set_title(metric); ax.set_xlabel("")
    plt.suptitle("CV metrics across folds")
    plt.tight_layout()
    plt.savefig(f"{out_dir}/cv_boxplot.png", dpi=120); plt.show()

    print("\n" + "=" * 70)
    print("Posterior diagnostics on full data (final models)")
    print("=" * 70)
    final_models = {}
    for target, config in FEAT_MAP.items():
        idata, scaler, post_sum = fit_full_and_diagnose(df, target, config, out_dir)
        final_models[target] = {"idata": idata, "scaler": scaler, "summary": post_sum}

    print(f"\nAll outputs saved to: {out_dir}")
    return fold_df, summary, thr_df, oof_predictions, p_machine, final_models


# =============================================================
# 8. 実行
# =============================================================
fold_df, summary, thr_df, oof_predictions, p_machine, final_models = main(csv_path)
```

# 予知保全モデル考察レポート

## 1. 概要

本レポートは、5つの故障モード（TWF、HDF、PWF、OSF、RNF）に対するベイズロジスティック回帰モデルの5分割層化交差検証結果を分析し、各モードのモデル性能および機械故障全体（OR結合）の予測性能を評価したものである。データは10,000サンプル、合計339件の故障（陽性率3.39%）を含み、各故障モードの陽性率は0.19%（RNF）から1.15%（HDF）まで幅がある不均衡データセットである。

## 2. 故障モード別の性能評価

### 2.1 完全予測達成モード（PWF、OSF）

PWFとOSFは AUC = 1.000、AP ≈ 1.000、F1 ≈ 0.99 と、ほぼ完全な予測性能を達成した。これは両モードがそれぞれ「`|Power - 6250| > 2750`」「`WearTorque > 型別閾値`」という1次元の決定論的閾値ルールに従うため、適切な特徴量エンジニアリング（PowerDev、WearTorque × Type ダミー）により線形分離可能な空間に変換できたことに起因する。Brier score も 0.0015、0.0016 と低く、確率の較正も良好である。Fold 間の標準偏差もほぼゼロであり、安定した予測が実現できている。実務上は閾値 0.38（PWF）、0.40（OSF）でアラートを発動すれば、FP は10,000サンプル中わずか1〜2件、FN はゼロとなり、完全自動化が可能な水準にある。

### 2.2 高精度だが改善余地のあるモード（HDF）

HDFは AUC = 0.9972、AP = 0.8165、F1 = 0.745、Brier = 0.0044 を達成した。閾値 0.32 において TP = 92、FP = 40、FN = 23 であり、115件中92件（80%）を検出できている。HDFの発生条件は「TempDiff < 8.6 K **かつ** rpm < 1380」という2次元のAND条件であり、線形ロジスティック回帰では完全分離が困難である。本モデルでは中心化交互作用項 `TempDiff × rpm` を導入し、事前分布で交互作用係数を正に設定することで、AND 条件を近似的に表現した。

ただしAP標準偏差が 0.0974 と他モードに比べて大きく、fold間のばらつきが認められる。これは2次元境界付近のサンプル分布が fold ごとに異なることに起因する統計的揺らぎである。理論上の上限は AUC = 1.000、AP = 1.000 であるため、B-spline 展開や明示的な閾値特徴量の導入により、さらに 0.18 程度の AP 改善余地が残されている。

### 2.3 確率的故障モード（TWF）

TWF は AUC = 0.9718 と高いランキング性能を示す一方、AP = 0.1413、F1 = 0.183 と低い識別性能にとどまった。閾値 0.084 において TP = 12、FP = 73、FN = 34 となり、46件中34件（74%）を検出できていない。これは故障の発生メカニズムが本質的に確率的であることに起因する。Tool wear が 200〜240分の窓内で 42.5% の確率で故障するという仕様であり、理論上のPrecision上限は約 0.425 に制約される。

現在の線形モデルは Tool wear に対して単調増加するシグモイド関数を出力するため、200分未満のサンプルにも不当に高い確率を割り当てている可能性があり、Precision が理論上限を大きく下回っている主因と考えられる。窓構造を正しく表現する Gaussian bump 型の非線形モデルや、明示的な範囲特徴量を導入することで、AP を 0.30〜0.40、F1 を 0.30 前後まで改善できる見込みがある。

### 2.4 純粋ランダムモード（RNF）

RNF は AUC = 0.500、AP = 0.0019（陽性率と一致）となり、説明変数からの予測が不可能であることを正しく反映している。本モデルではBeta-Bernoulli モデルにより全サンプルに同一の事後平均確率（≈ 0.0019）を割り当てる方針を採用しており、過学習を回避できている。Brier = 0.0019 は陽性率に基づくベースラインに等しく、これ以上の改善は構造上不可能である。実務的には機械学習ではなく定期点検等の運用で対応すべきモードである。

## 3. 機械故障全体（OR結合）の性能

5モードの予測確率を `p_machine = 1 - ∏(1 - p_mode)` で結合した結果、AUC = 0.9790、AP = 0.8389、Brier = 0.0119 を達成した。閾値 0.378 における混同行列は TP = 260、FP = 41、FN = 79、TN = 9620 であり、Precision = 0.864、Recall = 0.767、F1 = 0.812 となった。

実務的解釈としては、339件の真の故障のうち 260件（77%）を事前に検出でき、警告を発した301件のうち 86% が実際に故障する。全体に対する誤警報率は 41/9661 = 0.42% と十分低く、運用上許容できる水準である。一方、検出漏れ79件のうちその大半（34件）は TWF、残りは HDF（23件）、RNF（19件）等で構成されると推定され、今後の改善優先順位を示している。

## 4. 故障モードの分類と考察

本分析を通じて、故障モードは予測可能性の観点から以下の4類型に分類できることが明らかになった。第一の決定論的・線形分離可能型（PWF、OSF）は完全予測が達成され、実務的にも完全自動化に適する。第二の決定論的・非線形型（HDF）は高い予測性能を持つが、AND 条件等の非線形構造により完全予測には追加の特徴量工学が必要である。第三の確率的・部分予測可能型（TWF）はランキングは可能だが、本質的なランダム要素により Precision 上限が制約される。第四の純粋ランダム型（RNF）は機械学習では原理的に対応不可能であり、運用面で補完する必要がある。

この類型化は、各モードに対する適切なリソース配分の指針となる。決定論的モードに対しては自動アラートシステム、確率的モードに対してはリスクスコアに基づく優先点検、ランダムモードに対しては定期点検の組み合わせが推奨される。

## 5. ベイズアプローチの優位性

本モデルではベイズロジスティック回帰を採用した。これにより各予測に対して信頼区間（2.5%〜97.5%）が得られ、リスクベースの意思決定が可能となる点が主要な利点である。例えば、点予測ではなく信頼区間の下限値が一定値を超えた場合のみアラートを発する運用とすることで、より保守的かつ説明可能な判断が可能になる。また、物理的事前知識（HDFの交互作用項を正、TempDiff・rpmの主効果を負とする等）を事前分布として組み込むことで、線形モデルでありながら物理現象を反映した推論を実現している。

## 6. 改善優先順位と今後の課題

改善投資の優先順位は以下の通りである。最優先は TWF への非線形モデル（Gaussian bump 等）の導入であり、これにより機械故障全体の Recall を 0.77 から 0.82 程度まで押し上げられる見込みがある。次点は HDF の非線形特徴量追加で、AP を 0.95 程度まで引き上げる余地がある。一方、PWF・OSF はすでに上限に達しており追加投資は不要、RNF は機械学習による改善が原理的に不可能なため運用面での対応に委ねるべきである。

その他の課題として、現状のしきい値設定は F1 最大化基準であるが、実務では FN（見逃し）と FP（誤警報）のコストが非対称である場合が多い。コスト感応型しきい値の導入や、Precision ≥ 0.9 等の業務制約下での Recall 最大化も検討すべきである。また、calibration plot による較正性能の評価を全モードに拡張することで、確率予測の信頼性をさらに検証できる。

## 7. 結論

本ベイズロジスティック回帰モデルは、機械故障全体に対して AUC = 0.979、AP = 0.839、F1 = 0.812 という実用十分な予測性能を達成した。特に決定論的故障モード（PWF、OSF）では完全に近い予測を実現し、確率的・非線形モードについても物理知識に基づく特徴量工学と事前分布設定により高い性能を確保した。一方、TWF の非線形化と HDF のさらなる精緻化に改善余地があり、これらの実装により機械故障全体の検出率を大きく向上させる可能性がある。ベイズアプローチによる不確実性の定量化は、単なる点予測を超えたリスクベースの保全意思決定を可能とし、予知保全システムの実用化において重要な基盤となる。

```python
"""
点検優先度スコア出力
- p_machine_oof: 何らかの故障が起きる確率
- 各故障モード別の確率: どの故障が懸念されるか
- 優先度ランク: スコアが高い順
- リスクカテゴリ: しきい値ベースの分類
"""

import pandas as pd
import numpy as np

# 既存のoof_predictionsとp_machineから優先度スコアDFを構築
priority = pd.DataFrame({
    "UDI":             pd.read_csv(csv_path)["UDI"].values,     # 元データのID
    "Product ID":      pd.read_csv(csv_path)["Product ID"].values,
    "Type":            pd.read_csv(csv_path)["Type"].values,
    "p_machine":       p_machine,                               # 統合確率
    "p_TWF":           oof_predictions["TWF"]["mean"],
    "p_HDF":           oof_predictions["HDF"]["mean"],
    "p_PWF":           oof_predictions["PWF"]["mean"],
    "p_OSF":           oof_predictions["OSF"]["mean"],
    "p_RNF":           oof_predictions["RNF"]["mean"],
    "p_machine_lower": np.zeros(len(p_machine)),  # 後で埋める
    "p_machine_upper": np.zeros(len(p_machine)),
    "actual_failure":  pd.read_csv(csv_path)["Machine failure"].values,
})

# Machine failureの95%信用区間（各故障モードの下限・上限を伝播）
p_none_lo = np.ones(len(p_machine))
p_none_hi = np.ones(len(p_machine))
for t in ["TWF","HDF","PWF","OSF","RNF"]:
    p_none_lo *= (1.0 - oof_predictions[t]["upper"])  # 各上限の積 → 全体下限の逆
    p_none_hi *= (1.0 - oof_predictions[t]["lower"])
priority["p_machine_lower"] = 1.0 - p_none_hi
priority["p_machine_upper"] = 1.0 - p_none_lo

# 最も寄与している故障モードを特定
mode_probs = priority[["p_TWF","p_HDF","p_PWF","p_OSF","p_RNF"]].values
priority["dominant_mode"] = np.array(["TWF","HDF","PWF","OSF","RNF"])[mode_probs.argmax(axis=1)]
priority["dominant_prob"] = mode_probs.max(axis=1)

# リスクカテゴリ（4段階）
def categorize(p):
    if p >= 0.50: return "Critical"
    elif p >= 0.20: return "High"
    elif p >= 0.05: return "Medium"
    else: return "Low"
priority["risk_category"] = priority["p_machine"].apply(categorize)

# 優先度ランク（確率高い順）
priority = priority.sort_values("p_machine", ascending=False).reset_index(drop=True)
priority["priority_rank"] = priority.index + 1

# 保存
out_path = f"{OUT_DIR}/inspection_priority.csv"
priority.to_csv(out_path, index=False)
print(f"Saved: {out_path}")

# サマリー表示
print("\n" + "=" * 70)
print("Inspection Priority Summary")
print("=" * 70)
print(f"\nリスクカテゴリ別の件数:")
print(priority["risk_category"].value_counts().reindex(["Critical","High","Medium","Low"]))

print(f"\n優先度TOP20:")
cols = ["priority_rank","UDI","Product ID","Type","p_machine",
        "p_machine_lower","p_machine_upper","dominant_mode","risk_category","actual_failure"]
print(priority.head(20)[cols].to_string(index=False))

# 各リスクカテゴリ内での実際の故障率（モデル妥当性確認）
print(f"\n各カテゴリの実際の故障率（モデル妥当性チェック）:")
val = priority.groupby("risk_category").agg(
    n=("actual_failure","count"),
    actual_failures=("actual_failure","sum"),
    actual_failure_rate=("actual_failure","mean"),
    mean_predicted_prob=("p_machine","mean"),
).reindex(["Critical","High","Medium","Low"])
print(val.round(4))
```

# 点検優先度スコアリングシステム評価レポート

## 1. 概要

本レポートは、ベイズロジスティック回帰による予知保全モデルの予測結果を実務に活用するため、点検優先度スコアリングシステムを構築・評価した結果をまとめたものである。10,000サンプルの製造装置データに対して、機械故障全体の確率（p_machine）と5つの故障モード別確率、95%信用区間、支配的故障モード、リスクカテゴリを付与し、優先度順にランク付けした点検対象リストを生成した。本システムにより、限られた点検リソースを最もリスクの高い装置に集中させる意思決定支援が可能となる。

## 2. システムの構成

本システムは5つの構成要素から成る。第一に各サンプルへの統合確率と故障モード別確率の付与、第二に95%信用区間の伝播計算による不確実性の定量化、第三に支配的故障モードの自動特定、第四にリスクカテゴリへの4段階分類、第五に優先度ランクの付与である。

信用区間の伝播は、各故障モードの上限・下限を用いた `p_machine = 1 - ∏(1 - p_mode)` の形で計算した。これは故障モード間の独立性を仮定した近似だが、各モードが異なる物理メカニズムに基づくため実務上妥当な近似である。リスクカテゴリは p_machine の値に基づき、0.50以上を Critical、0.20以上を High、0.05以上を Medium、それ未満を Low に分類した。これらの閾値は F1 最適化しきい値（0.378）とは別に、実務での点検リソース配分を意識した粗い区分として設定した。

## 3. リスクカテゴリ別の分布

10,000サンプルの分類結果は Critical 232件、High 210件、Medium 462件、Low 9,096件となった。Critical と High を合わせた高リスク群は全体の4.42%（442件）であり、これが優先的に点検すべき対象となる。Medium を含めた要監視対象は904件（9.04%）、残り約91%は通常運用可能と判断できる。

この分布は、ヒトの点検能力に対して適切な絞り込みが実現されていることを示している。仮に1日あたり50台の点検能力があれば、Critical群は約4.6日、High群まで含めても約9日で網羅でき、現実的な運用が可能な水準である。実故障の事前確率（3.39%）に対して、Critical 群の故障率は約27倍に濃縮されており、点検効率の大幅な向上が期待できる。

優先度TOP20の結果は極めて高い予測精度を示した。20件すべてで `actual_failure = 1` となり、Precision = 100% を達成した。p_machine の値はすべて 0.999 以上で、信用区間の下限も 0.997 以上と非常に狭く、モデルが高い確信を持って故障を予測していることが確認できる。

支配的故障モード（dominant_mode）の内訳は PWF が12件、OSF が5件、HDF が2件、PWF（残り1件）と多様性が認められる。これは前回（PWF 12件、OSF 8件のみ）と比較して HDF が上位に進出している点が注目される。PWF と OSF の予測モデルが AUC = 1.000、AP ≈ 1.000 と完全に近い性能を持つため最も高信頼な予測が上位を占める一方、HDF についても交互作用項の改善により AUC = 0.997、AP = 0.82 へ向上し、上位に食い込めるレベルの確信度が得られていることを示す。一方、TWF と RNF はそれぞれ確率的故障とランダム故障のため、上位20件には含まれない。

具体的には、TWF に対しては Tool wear が 180 分を超えた装置を定期点検対象とする時間ベースのルール、RNF に対しては定期巡回点検が望ましい。これらをシステムと併用することで、機械学習で対応可能な故障モードと運用ルールでカバーすべき故障モードを役割分担できる。

## 9. 前回との比較と改善効果

HDF モデルの非線形化（交互作用項追加）により、本システムには以下の改善が現れている。リスクカテゴリの分布は Critical 221件 → 232件（+11件）、High 233件 → 210件（-23件）と再配分が起き、より高確信のサンプルが Critical に格上げされた。Critical の実故障率も 90.50% → 91.81% へ向上し、Precision が改善している。TOP20には HDF を支配モードとするサンプルが新たに2件登場し、HDF の識別能力向上が直接的な貢献として現れている。

## 10. 結論

本点検優先度スコアリングシステムは、ベイズロジスティック回帰の予測結果を実用的な意思決定支援ツールに変換することに成功した。Critical カテゴリの 232件で実故障の62.8%を捕捉する高い濃縮効率、High まで含めた442件で総故障の81.1%を捕捉する優れた検出能力、各リスクカテゴリで良好に較正された予測確率、そして信用区間による不確実性の定量化を併せ持つことで、限られた点検リソースの最適配分を支援する実用的なシステムとなっている。Critical カテゴリでの保守的な過小予測は安全側の誤差として実務的に望ましく、信頼できるアラートシステムとして運用可能な水準に達している。今後の改善方向として、TWF の非線形モデル化による Medium カテゴリの精度向上、信用区間の下限を活用した保守的しきい値の導入、コスト感応型のしきい値最適化等が挙げられ、これらの実装によりさらなる実務的価値の向上が期待できる。
